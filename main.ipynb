```python
hdi = pd.read_csv(R"C:\Users\TAJ\Desktop\World Economic Report\HDI.csv")
wb_hdi_2014 = df.query("Year == 2014").merge(hdi[['iso3','hdi_2014']],how='left', left_on="Country Code", right_on="iso3")
wb_hdi_2014.head()
```

```python
gdp_pivot = df.pivot_table(index="Year", columns="Region", values="GDP (USD)", aggfunc="sum")
gdp_pivot.head()
```

```python
pop_pivot = df.pivot_table(index="Year", columns="Region", values="Population (M)", aggfunc="sum")
pop_pivot.head()
```
