```python
# Prepare for visualization
```

```python
gdp_pivot = df.pivot_table(index="Year", columns="Region", values="GDP (USD)", aggfunc="sum")
gdp_pivot.head()
```

```python
pop_pivot = df.pivot_table(index="Year", columns="Region", values="Population (M)", aggfunc="sum")
pop_pivot.head()
```

```python
df.columns = df.columns.str.strip()
```

```python
df.columns
```

```python
wb_hdi_by_region = wb_hdi_2014.groupby(
    "Region"
).agg({"hdi_2014": "mean"}).sort_values("hdi_2014", ascending=False)
wb_hdi_by_region.head()
```

```python
# Visualize the Data
```

```python
# Stack area
fig, ax = plt.subplots()

ax.stackplot(
    gdp_pivot.index,
    [gdp_pivot[region] / 1_000_000_000_000 for region in gdp_pivot.
     iloc[-1].sort_values(ascending=False).index],
     labels=gdp_pivot.iloc[-1].sort_values(ascending=False).index
)

ax.legend(loc="upper left")
plt.title("GDP has Grows Exponentially Over Time")
ax.set_ylabel("GDP (Trillions)")
plt.show()
```

```python
gdp_pivot.iloc[-1].sort_values(ascending=False).index
```

```python
# Stack area
fig, ax = plt.subplots()

ax.stackplot(
    pop_pivot.index,
    [pop_pivot[region] / 1_000 for region in pop_pivot.
     iloc[-1].sort_values(ascending=False).index],
     labels=pop_pivot.iloc[-1].sort_values(ascending=False).index
)

ax.legend(loc="upper left")
plt.title("Population has surged from 2 billion to 7.5 billions")
ax.set_ylabel("Population (Trillions)")
plt.show()
```
